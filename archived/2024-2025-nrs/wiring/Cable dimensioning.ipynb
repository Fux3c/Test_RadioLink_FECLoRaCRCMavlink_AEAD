## Voltage drop
Voltage drop across cables calculated from the following derived formula:

$$ A = \frac{2\times \rho(T) \times L \times I}{\Delta V} $$
Where $L$ is the length of the cable in meters, $\Delta V$ is the accepted voltage drop, and $\rho(T)$ is:
$$ \rho(T) = \rho _0 \times[1+\alpha\times(T-T_0)] $$
Here $\rho _0$ is the material resistivity, $\alpha$ is the temperature coefficient for the material and $T_0$ is the reference temperature for the material's resisitivity.(I.E. the temperature where $\rho$ of the material is equal to $\rho_ 0$)

```python
copper = {
    'resistivity': 1.68e-8,  # ohm-meters
    'temperature_coefficient': 0.0039,  # percentage per degree Celsius
    'reference_temperature': 20  # degrees Celsius for the resistivity value

}
def calculate_wire_crossection(length: float, 
                               current: float, 
                               voltage_drop: float, 
                               temperature: float, 
                               material: dict = copper)-> float:
    """
    Calculate the required cross-sectional area of a wire.

    Parameters:
    - length (float): Length of the wire in meters.
    - current (float): Current flowing through the wire in amperes.
    - voltage_drop (float): Permissible voltage drop in volts.
    - temperature (float): Operating temperature in degrees Celsius.
    - material (dict): Material properties of the wire.
    Returns:
    - float: Required cross-sectional area in square millimeters.
    """

    # Calculate resistivity at the given temperature
    resistivity_at_temp = material['resistivity'] * (1 + material['temperature_coefficient'] * (temperature - material['reference_temperature']))
    
    # Calculate the required cross-sectional area in square meters
    area_in_square_meters = (2 * resistivity_at_temp * length * current) / voltage_drop
    
    # Convert the area to square millimeters (1 m^2 = 1e6 mm^2)
    area_in_square_mm = area_in_square_meters * 1e6
    
    return area_in_square_mm
```

## HPRS
Servo 

Setting the acceptable voltage drop $\Delta V$ to $0.2V$, $L = 2.5M$, $I_{Peak} = 2A$ and $T = 60^{\circ}C$ gives us: 


```python
length = 1.5  # in meters
length = length + 2*0.1 # add 10 cm on each side for connections and service loops
current = 3  # in amperes
voltage_drop = 0.2  # in volts
temperature = 70  # in degrees Celsius
required_area = calculate_wire_crossection(length, current, voltage_drop, temperature, material=copper)
print(f"Required cross-sectional area: {required_area:.2f} mm²")
```
